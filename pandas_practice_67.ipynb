## Level 1 — Exercise

Use `sns.load_dataset('exercise')`. Group by `diet` and `kind`.

Build a named-agg summary with `avg_pulse`, `std_pulse`, `n`, and `pulse_range` (max minus min).

- Which (`diet`, `kind`) combination produces the highest average pulse?
- Does higher average pulse tend to come with higher variability? Print the correlation value and give a one-sentence interpretation.
- Without a loop, find which exercise `kind` has the most consistent pulse across diet groups.

```python
exercise = sns.load_dataset('exercise')

# Your code here
g = exercise.groupby(['diet','kind'], observed=True).agg(
    avg_pulse = ('pulse','mean'),
    std_pulse = ('pulse','std'),
    n = ('pulse','count'),
    pulse_range = pd.NamedAgg(column='pulse', aggfunc=(lambda x: x.max() - x.min()))
    
)

print(g['avg_pulse'].idxmax(),'produces the highest average pulse')

np.corrcoef(g['avg_pulse'],g['std_pulse'])[0,1]
print('highly correlated')

print(g['avg_pulse'].unstack().std(axis=0).idxmin(),'has the most consistent pulse across diet groups')

```

## Level 2 — Product orders

`orders` records individual purchases across four product categories and two sales channels. Add `net_revenue = price * quantity * (1 - discount)` before grouping.

Group by `category` and `channel`. Named agg: `avg_net_rev` (mean of `net_revenue`), `total_orders` (count), `avg_discount` (mean of `discount`), `avg_qty` (mean of `quantity`).

- Which (`category`, `channel`) pair has the highest average net revenue?
- Is there a tradeoff between `avg_discount` and `avg_net_rev` across groups? Print the correlation value.
- What are the 25th, 50th, and 75th percentiles of `avg_net_rev` across all groups?

```python
np.random.seed(22)
m = 180
orders = pd.DataFrame({
    'category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Books'], m),
    'channel':  np.random.choice(['Online', 'In-Store'], m),
    'quantity': np.random.randint(1, 10, m),
    'price':    np.random.choice([9.99, 19.99, 39.99, 79.99, 149.99], m),
    'discount': np.random.choice([0, 0.05, 0.1, 0.15, 0.2], m),
})

# Your code here
orders['net_revenue'] = orders['price'] * orders['quantity']*(1-orders['discount'])

g =orders.groupby(['category','channel']).agg(
    avg_net_rev = ('net_revenue','mean'),
    total_orders = ('price','count'),
    avg_discount = ('discount','mean'),
    avg_qty = ('quantity','mean')
)

print(g['avg_net_rev'].idxmax(),'has the highest avg net rev')

cc = np.corrcoef(g['avg_discount'],g['avg_net_rev'])[0,1]
print(f'correlation is {cc:.2f}')
print('not very correlated')

print(np.percentile(g['avg_net_rev'], q = [25,50,75]))
```

## Level 3 — Attention

Use `sns.load_dataset('attention')`. The dataset has `attention` (focused / divided), `solutions` (1, 2, or 3), and `score`.

Group by `attention` and `solutions`. Build a named-agg summary with `avg_score`, `std_score`, and `n`. No hints.

- Which (`attention`, `solutions`) combination produces the highest average score?
- Which attention type has the higher average score overall — without using a loop?
- Does score variability (`std_score`) correlate with average score across the six groups? Print the value.

```python
attention = sns.load_dataset('attention')

# Your code here


g = attention.groupby(['attention','solutions']).agg(
    avg_score = ('score','mean'),
    std_score = ('score','std'),
    n = ('score','count')
)

print(g['avg_score'].idxmax(),'has the highest average score')

print(g['avg_score'].unstack().mean(axis=1).idxmax(), 'has the higher score overall')

cc= np.corrcoef(g['std_score'], g['avg_score'])[0,1]
print('correlation is ', cc)
print('negatively correlated')
```
