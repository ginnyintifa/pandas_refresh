## Level 2 — Equipment failures & maintenance

**New concept: `direction='forward'` in `merge_asof`**

The default `direction='backward'` finds the most recent *past* value. `direction='forward'` flips this: for each row in `left`, it finds the smallest key in `right` that is **≥** the left key — the *next* future value.

```python
pd.merge_asof(left, right, on='date', direction='forward')
```

If no future match exists (the left key is past the end of `right`), NaN is returned.
