## Level 1 — Marathon checkpoints

`checkpoints` logs temperature and humidity at fixed intervals throughout the race. `runners` records when each person crossed the finish area.

1. Sort both by `time` and merge with `merge_asof`.
2. Add `heat_index = temperature + humidity * 0.1`. Which runner faced the worst conditions? Use `np.argmax`.
3. Named-agg groupby by `category`: mean `heat_index` and mean `temperature`.
4. Use `np.argsort` on the mean heat indices to rank categories best → worst conditions.

````python
checkpoints = pd.DataFrame({
    'time':        pd.to_datetime(['08:00','08:45','09:30','10:15','11:00','11:30'], format='%H:%M'),
    'temperature': [62, 65, 70, 76, 79, 81],
    'humidity':    [55, 58, 62, 68, 72, 75],
})

runners = pd.DataFrame({
    'time':     pd.to_datetime(['08:12','08:55','09:08','09:47','10:30','11:18'], format='%H:%M'),
    'name':     ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank'],
    'category': ['elite', 'amateur', 'elite', 'amateur', 'elite', 'amateur'],
})

# Your code here

checkpoints = checkpoints.sort_values('time')
runners = runners.sort_values('time')

m = pd.merge_asof(
    runners, 
    checkpoints,
    on='time'
)
m['heat_index'] = m['temperature'] + m['humidity'] *0.1
print(m.iloc[np.argmax(m['heat_index'])]['name'],'faced the worst conditions')

g = m.groupby('category').agg(
    mean_hi = ('heat_index','mean'),
    mean_temp = ('temperature','mean')
)

print(g.index[np.argsort(g['mean_hi'])])
````

## Level 2 — Equipment failures & maintenance

**New concept: `direction='forward'` in `merge_asof`**

The default `direction='backward'` finds the most recent *past* value. `direction='forward'` flips this: for each row in `left`, it finds the smallest key in `right` that is **≥** the left key — the *next* future value.

```python
pd.merge_asof(left, right, on='date', direction='forward')
```

If no future match exists (the left key is past the end of `right`), NaN is returned.

`failures` logs equipment breakdowns. `schedule` lists upcoming maintenance windows. For each failure, find the *next* scheduled maintenance.

1. Sort both by `date`. Merge with `direction='forward'`.
2. How many failures have no upcoming maintenance? Print their dates.
3. Use `np.nanmean` on `duration` to find the average maintenance length tied to these failures.
4. Among matched failures only, which one is linked to the longest maintenance? Find the failure date.

````python
failures = pd.DataFrame({
    'date':     pd.to_datetime(['2023-01-15','2023-03-20','2023-05-08','2023-07-22','2023-09-14','2023-12-10']),
    'severity': ['low', 'high', 'medium', 'low', 'high', 'medium'],
})

schedule = pd.DataFrame({
    'date':      pd.to_datetime(['2023-02-01','2023-04-15','2023-06-01','2023-08-01','2023-10-01','2023-11-15']),
    'maint_type': ['inspection', 'repair', 'inspection', 'repair', 'inspection', 'repair'],
    'duration':  [2, 5, 2, 4, 2, 6],
})

# Your code here


failures = failures.sort_values('date')
schedule = schedule.sort_values('date')

m = pd.merge_asof(failures, 
                  schedule, 
                  on = 'date',
                  direction='forward')

mn = m.loc[m['maint_type'].isna(),'date']
print(mn.values, 'has no upcoming maintenance')
print(len(mn),'has no upcoming maintenance')


print(np.nanmean(m['duration']))

mmm = m.dropna()

print(mmm.iloc[np.argmax(mmm['duration'])]['date'],'is linked to the longest')


````

## Level 3 — Bond yield spread

`govt` and `corp` report weekly bond yields, but on alternating weeks. Build a complete weekly view and analyze the credit spread (the gap between corporate and government yields).

1. Use `merge_ordered` with `fill_method='ffill'`. Drop rows still NaN after the fill.
2. Add `credit_spread = corp_yield - govt_yield`. Which week had the widest spread? Which had the narrowest? (Use `.idxmax()` / `.idxmin()` and then `.loc[]` to get the actual date.)
3. Use `np.corrcoef` on the two yield columns.
4. Use `np.percentile` to find the 25th and 75th percentiles of the credit spread.
5. Use a list comprehension to collect weeks where `credit_spread > 1.50`.

````python
govt = pd.DataFrame({
    'week':       pd.to_datetime(['2023-01-02','2023-01-16','2023-01-30','2023-02-13','2023-02-27','2023-03-13']),
    'govt_yield': [3.82, 3.91, 4.05, 4.12, 4.08, 3.97],
})

corp = pd.DataFrame({
    'week':       pd.to_datetime(['2023-01-09','2023-01-23','2023-02-06','2023-02-20','2023-03-06','2023-03-20']),
    'corp_yield': [5.41, 5.38, 5.52, 5.47, 5.61, 5.55],
})

# Your code here



m = pd.merge_ordered(
    govt, 
    corp, 
    on = 'week',
    fill_method='ffill'
).dropna()

m['credit_spread'] = m['corp_yield'] - m['govt_yield']


print(m.loc[m['credit_spread'].idxmax(),'week'],'has the widest spread')

print(m.loc[m['credit_spread'].idxmin(),'week'],'has the narrowest spread')

print(np.corrcoef(m['govt_yield'], m['corp_yield'])[0,1])



print(np.percentile(m['credit_spread'], q = [25,75]))


[w for w, s in zip(m['week'],m['credit_spread']) if s >1.50]


````
