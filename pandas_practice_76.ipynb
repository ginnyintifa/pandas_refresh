## Level 1 — Monthly revenue

**New concept: `expanding()`**

`expanding()` creates a window that starts at the first row and grows to include every row up to the current one:

```python
df['col'].expanding().mean()   # cumulative mean up to each row
df['col'].expanding().std()    # cumulative standard deviation
df['col'].expanding().max()    # running maximum
```

Unlike `rolling(n)` which uses a fixed window of the last *n* rows, `expanding()` always uses all data up to that point. `.mean()` and `.max()` are valid from the first row. `.std()` requires at least 2 rows — the first row returns NaN.

Monthly revenue for an online store. Use `expanding()` to track how performance looks relative to the running history.

1. Add `cum_mean = amount.expanding().mean()` and `cum_max = amount.expanding().max()`.
2. Which months had revenue **above** the cumulative mean at that point? Use a boolean mask.
3. Which months set a new all-time record (`amount == cum_max`)?
4. Add `zscore = (amount - cum_mean) / amount.expanding().std()`. Which month had the highest z-score? Use `np.argmax` — skip the first row where std is NaN.

````python
revenue = pd.DataFrame({
    'month':  pd.to_datetime(['2023-01','2023-02','2023-03','2023-04','2023-05',
                              '2023-06','2023-07','2023-08','2023-09','2023-10'], format='%Y-%m'),
    'amount': [42000, 38000, 45000, 51000, 47000, 55000, 49000, 62000, 58000, 71000],
})

# Your code here

revenue['cum_mean'] = revenue['amount'].expanding().mean()
revenue['cum_max'] = revenue['amount'].expanding().max()

print(revenue.loc[revenue['amount']>revenue['cum_mean'],'month'])

print(revenue.loc[revenue['amount']==revenue['cum_max'],'month'])

revenue['zscore'] = (revenue['amount'] - revenue['cum_mean'])/ (revenue['amount'].expanding().std())

mn = revenue.dropna()
print(mn.iloc[np.argmax(mn['zscore'])]['month'], 'had the highest z-score')
````

## Level 2 — Sensor anomaly detection

A temperature sensor records weekly readings. Use `expanding()` to build dynamic bounds — instead of comparing against a fixed threshold, flag readings that fall more than 2 standard deviations from the cumulative mean **up to that point**. This means the bounds tighten or widen as more data arrives.

1. Add `cum_mean` and `cum_std` using `expanding()`.
2. Add `upper = cum_mean + 2 * cum_std`. For the first row, `cum_std` is NaN so `upper` is also NaN — no flagging possible yet.
3. Add `anomaly = temp_c > upper`. How many anomalies? Which timestamps?
4. For non-anomalous readings only: use `np.nanmean` to find the average temperature. Compare to `np.mean` on all readings — how much do the anomalies skew the overall mean?

````python
sensors = pd.DataFrame({
    'timestamp': pd.date_range('2023-01-01', periods=12, freq='W'),
    'temp_c':    [20.1, 21.3, 19.8, 22.5, 20.9, 21.7, 23.4, 20.2, 24.8, 21.1, 19.5, 28.6],
})

# Your code here

sensors['cum_mean'] = sensors['temp_c'].expanding().mean()
sensors['cum_std'] = sensors['temp_c'].expanding().std()

sensors['upper'] = sensors['cum_mean'] + 2 *sensors['cum_std']
sensors['anomaly'] = sensors['temp_c'] > sensors['upper']
print(sensors['anomaly'].sum(),'are anomalies')

print(sensors.loc[sensors['anomaly'],'timestamp'])

avt = np.nanmean(sensors.loc[~sensors['anomaly'],'temp_c'])
a = np.mean(sensors['temp_c'])


print(a-avt, 'is the skew')
````

## Level 3 — Two sales channels

`online` and `instore` report weekly sales on alternating weeks. Merge them into a full timeline, then use `expanding()` to track cumulative performance.

1. `merge_ordered` + ffill + dropna.
2. Add `total = online_sales + instore_sales`. Compute `cum_total = total.expanding().sum()` — how much revenue has been earned in total up to each week?
3. Add `cum_avg = total.expanding().mean()`. Which weeks had total above the running average?
4. `np.corrcoef` on `online_sales` and `instore_sales` — do the two channels move together?
5. Use `np.percentile` on `total` to find the 25th and 75th percentiles. How many weeks fall in the top quartile?

````python
online = pd.DataFrame({
    'week':         pd.to_datetime(['2023-01-02','2023-01-16','2023-01-30','2023-02-13','2023-02-27']),
    'online_sales': [8500, 9200, 8800, 10100, 9700],
})

instore = pd.DataFrame({
    'week':          pd.to_datetime(['2023-01-09','2023-01-23','2023-02-06','2023-02-20','2023-03-06']),
    'instore_sales': [12400, 11800, 13200, 12600, 14100],
})

# Your code here


m = pd.merge_ordered(
    online, 
    instore, 
    on = 'week',
    fill_method='ffill'
).dropna()

m['total'] = m['online_sales'] + m['instore_sales']
m['cum_total']= m['total'].expanding().sum()
m['cum_avg'] = m['total'].expanding().mean()

print(m.loc[m['total']>m['cum_avg'],'week'],' had total above running averag')


cor = np.corrcoef(m['online_sales'],m['instore_sales'])[0,1]
print(f'correlation is {cor:.2f}')
print('move together')
qs = np.percentile(m['total'],q =[25,75])
print((m['total']>qs[1]).sum(),'fall in top quartile')
````
