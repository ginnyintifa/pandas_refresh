## Level 1 — Monthly revenue

**New concept: `expanding()`**

`expanding()` creates a window that starts at the first row and grows to include every row up to the current one:

```python
df['col'].expanding().mean()   # cumulative mean up to each row
df['col'].expanding().std()    # cumulative standard deviation
df['col'].expanding().max()    # running maximum
```

Unlike `rolling(n)` which uses a fixed window of the last *n* rows, `expanding()` always uses all data up to that point. `.mean()` and `.max()` are valid from the first row. `.std()` requires at least 2 rows — the first row returns NaN.
