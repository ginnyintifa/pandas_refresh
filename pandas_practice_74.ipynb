## Level 1 — Drug dosing protocols

`protocols` logs dosing changes throughout the day. `vitals` records patient measurements at irregular times. For each vital reading, attach the dosing protocol in effect at that moment.

1. Sort and merge.
2. Add `response_ratio = heart_rate / dose_mg` — how much heart rate per mg of medication. Higher means the drug is less effective per mg. Which reading had the worst (highest) ratio? Use `np.argmax`.
3. Named-agg groupby by `protocol`: mean `heart_rate` and mean `response_ratio`.
4. Use `np.argsort` to rank protocols from highest to lowest mean response ratio.

```python
protocols = pd.DataFrame({
    'time':     pd.to_datetime(['06:00','10:00','14:00','18:00'], format='%H:%M'),
    'dose_mg':  [5.0, 7.5, 5.0, 10.0],
    'protocol': ['low', 'medium', 'low', 'high'],
})

vitals = pd.DataFrame({
    'time':       pd.to_datetime(['07:30','09:15','11:45','13:00','15:30','19:45'], format='%H:%M'),
    'patient_id': ['A', 'B', 'A', 'B', 'A', 'B'],
    'heart_rate': [78, 92, 85, 110, 88, 95],
})

# Your code here
protocols = protocols.sort_values('time')
vitals = vitals.sort_values('time')

m = pd.merge_asof(vitals, 
             protocols, 
             on = 'time')

m['response_ratio'] = m['heart_rate']/m['dose_mg']
print(m.iloc[np.argmax(m['response_ratio'])]['time'],'had the worst')

g = m.groupby('protocol').agg(
    m_h = ('heart_rate','mean'),
    m_r = ('response_ratio','mean')
)
print(g)
print(
    g.index[np.argsort(-g['m_r'])]
)
```

## Level 2 — Appointment booking

`slots` lists available appointment windows. `requests` records when patients asked for an appointment. For each request, find the **next available slot within 7 days**. Requests with no slot in that window get NaN.

Remember: sort both by `date` only (not by a group column) before merging.

1. Merge with the correct direction and tolerance.
2. How many requests received no slot? Print their patient names and urgency levels.
3. Use `np.nanmean` on `duration` across all requests.
4. Among matched requests only, which patient has the longest slot? Use `np.argmax`.

```python
slots = pd.DataFrame({
    'date':     pd.to_datetime(['2023-07-03','2023-07-05','2023-07-08',
                                '2023-07-12','2023-07-15','2023-07-18']),
    'type':     ['general','specialist','general','specialist','general','specialist'],
    'duration': [30, 60, 30, 60, 30, 60],
})

requests = pd.DataFrame({
    'date':     pd.to_datetime(['2023-07-01','2023-07-06','2023-07-09',
                                '2023-07-16','2023-07-19','2023-07-25']),
    'patient':  ['Alpha', 'Beta', 'Gamma', 'Delta', 'Echo', 'Foxtrot'],
    'urgency':  ['high', 'low', 'high', 'low', 'high', 'low'],
})

# Your code here

slots = slots.sort_values('date')
requests = requests.sort_values('date')

m = pd.merge_asof(
    requests, 
    slots,
    on = 'date',
    tolerance=pd.Timedelta('7D'),
    direction='forward'
)
print(m['duration'].isna().sum(),'has NAN')
print(m.loc[m['duration'].isna(),'patient'])
print(m.loc[m['duration'].isna(),'urgency'])

print(np.nanmean(m['duration']))

mn = m.dropna()
print(mn.iloc[np.argmax(mn['duration'])]['patient'],'has the longest slot')
```

## Level 3 — Employment and inflation

`employment` and `inflation` track economic indicators on alternating quarters. Neither has a complete record.

Build a full quarterly view and answer the following — less hand-holding this time:

1. `merge_ordered` + ffill + dropna.
2. Add `pressure = cpi_change / (100 - emp_rate)`. This is inflation per unit of unemployment — higher means prices are rising fast despite low joblessness. Which quarter had the highest pressure?
3. `np.corrcoef` on the two columns — do employment and inflation tend to move together or in opposite directions in this data?
4. Compute each metric's mean with `np.mean`. Then find the "sweet spot" quarters where employment is above its mean AND inflation is below its mean — use a list comprehension with `zip` over three columns.

```python
employment = pd.DataFrame({
    'date':     pd.to_datetime(['2022-01','2022-07','2023-01','2023-07'], format='%Y-%m'),
    'emp_rate': [95.2, 96.8, 95.9, 97.4],
})

inflation = pd.DataFrame({
    'date':       pd.to_datetime(['2022-04','2022-10','2023-04','2023-10'], format='%Y-%m'),
    'cpi_change': [3.2, 4.1, 3.8, 2.9],
})

# Your code here
m = pd.merge_ordered(
    employment, 
    inflation, 
    on = 'date',
    fill_method='ffill').dropna()

m['pressure'] = m['cpi_change']/(100 - m['emp_rate'])
print(m.loc[m['pressure'].idxmax(),'date'],'has the highest pressure')

c = np.corrcoef(m['emp_rate'],m['cpi_change'])[0,1]

print("correlation is: ",c)

print('slight negative correlation')
em = np.mean(m['emp_rate'])
cm = np.mean(m['cpi_change'])

[q for q, e, c in zip(m['date'],m['emp_rate'],m['cpi_change']) if (e > em) & (c <cm)]
```
