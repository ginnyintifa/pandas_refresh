## Level 1 — Electricity prices

**New concept: `diff()`**

`diff()` gives the absolute change from the previous row — current minus previous:

```python
s.diff()     # current - previous  (absolute change, same units as s)
s.diff(n)    # current - n periods ago
```

Compare to `pct_change()` which gives `(current - previous) / previous` — a relative change. Use `diff()` when you care about magnitude in original units; use `pct_change()` when you care about rate of change.

Both return `NaN` for the first row.
