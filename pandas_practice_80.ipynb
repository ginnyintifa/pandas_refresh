## Level 1 — Electricity prices

**New concept: `diff()`**

`diff()` gives the absolute change from the previous row — current minus previous:

```python
s.diff()     # current - previous  (absolute change, same units as s)
s.diff(n)    # current - n periods ago
```

Compare to `pct_change()` which gives `(current - previous) / previous` — a relative change. Use `diff()` when you care about magnitude in original units; use `pct_change()` when you care about rate of change.

Both return `NaN` for the first row.

Monthly electricity price ($/kWh) over two years.

1. Add `abs_change = price_kwh.diff()` and `rel_change = price_kwh.pct_change()`. Print rows 0–5 of all three columns. In which months do they tell a different story about what happened?
2. `np.nanmean` and `np.nanstd` on both `abs_change` and `rel_change`. What's the typical monthly change in dollars vs percentage?
3. Add `ewm_change = abs_change.ewm(span=6).mean()`. Which month had the most extreme smoothed change (positive or negative)? Use `np.nanargmax` on `np.abs(ewm_change)`. Print that month and the ewm_change value — what does it mean?

````python
prices = pd.DataFrame({
    'month':     pd.date_range('2021-01', periods=24, freq='MS'),
    'price_kwh': [0.112, 0.108, 0.115, 0.119, 0.124, 0.118,
                  0.122, 0.127, 0.131, 0.138, 0.142, 0.135,
                  0.141, 0.148, 0.155, 0.163, 0.171, 0.168,
                  0.175, 0.182, 0.178, 0.185, 0.191, 0.188],
})

# Your code here

prices['abs_change'] = prices['price_kwh'].diff()
prices['rel_change'] = prices['price_kwh'].pct_change()

print(prices.iloc[:5][['price_kwh','abs_change','rel_change']])

abs_mean = np.nanmean(prices['abs_change'])
rel_mean = np.nanmean(prices['rel_change'])

abs_std = np.nanstd(prices['abs_change'])
rel_std = np.nanstd(prices['rel_change'])

print(abs_mean, abs_std, rel_mean, rel_std)

prices['ewm_change'] = prices['abs_change'].ewm(span = 6).mean()


print('month is: ', prices.iloc[np.nanargmax(np.abs(prices['ewm_change']))]['month'])
print('value is: ', prices.iloc[np.nanargmax(np.abs(prices['ewm_change']))]['ewm_change'])
````

## Level 3 — Three retail branches

Weekly sales for three branches over 12 weeks. Each has a distinct trajectory.

1. Add `wow_change` (week-over-week absolute change) via `groupby('branch').transform(lambda x: x.diff())`. Which branch × week had the sharpest single-week drop?
2. Add `ewm_sales` via `groupby('branch').transform(lambda x: x.ewm(span=4).mean())`. At week 12, use `np.argsort` on the three final ewm values to rank branches. Which is trending strongest?
3. Extract each branch's `sales` as a plain array. Use `np.corrcoef` + `np.fill_diagonal` + `np.unravel_index` to find the most-correlated pair of branches.
4. `np.percentile` on `ewm_sales` across all rows — find Q1. Which branch spends the most weeks below it? What does that say about that branch?

````python
branches = pd.DataFrame({
    'week':   list(pd.date_range('2023-01-02', periods=12, freq='W')) * 3,
    'branch': ['Alpha']*12 + ['Beta']*12 + ['Gamma']*12,
    'sales':  [
        24000, 25200, 24800, 26100, 27300, 26800, 28200, 29100, 28600, 30200, 31500, 32800,  # Alpha
        38000, 35200, 41000, 36800, 39500, 37200, 40800, 36500, 39200, 37800, 40500, 38100,  # Beta
        52000, 50800, 51500, 49200, 48700, 47100, 48900, 46200, 45800, 44100, 43700, 42200,  # Gamma
    ],
})

# Your code here

branches['wow_change'] = branches.groupby('branch')['sales'].transform(lambda x: x.diff())
print(branches.iloc[np.nanargmin(branches['wow_change'])]['branch'])
print(branches.iloc[np.nanargmin(branches['wow_change'])]['week'])

branches['ewm_sales'] = branches.groupby('branch')['sales'].transform(lambda x: x.ewm(span = 4).mean())

ws = branches['week'].unique()
b12 = branches.loc[branches['week'] == ws[11]]
b12_sorted = b12.iloc[np.argsort(-b12['ewm_sales'])]
print(b12_sorted)
print(b12_sorted.iloc[0]['branch'],'is trending strongest')


sa = np.array(branches.loc[branches['branch'] == 'Alpha','sales'])
sb = np.array(branches.loc[branches['branch'] == 'Beta','sales'])
sg = np.array(branches.loc[branches['branch'] == 'Gamma','sales'])

cc = np.corrcoef([sa,sb,sg])
np.fill_diagonal(cc, -np.inf)
flat_index = np.argmax(cc)
row,col = np.unravel_index(flat_index, cc.shape)

bn = ['Alpha','Beta','Gamma']
print('most correlated pair is: ',bn[row], bn[col])

q = np.percentile(branches['ewm_sales'], q = [25])

gg = branches.groupby('branch')['ewm_sales'].apply(lambda x: (x<q[0]).sum())
print(gg.idxmax(),'spends most weeks under Q1')
print('sales not good')
````
