## Level 1 — Coffee shop orders

A café exported a morning's orders. Clean it and answer the questions — no hints.

- Which item was ordered most? Which size?
- What is the average price per item type?
- What fraction of orders used a loyalty card?

```python
orders = pd.DataFrame({
    'order_id': ['C01','C02','C03','C04','C05','C06','C07','C02'],
    'item':     ['Latte','Espresso','Cold Brew','LATTE','Cold Brew','espresso','Latte','Espresso'],
    'size':     ['Large','Small','Medium','large','MEDIUM','small','Large','Small'],
    'price':    ['$5.50','$3.00','$4.75','$5.50','$4.75','$3.00','$5.50','$3.00'],
    'loyalty':  ['Yes','No','Yes','YES','no','No','Yes','No'],
})

# Your code here

orders['item'] = orders['item'].str.lower()
orders['size'] = orders['size'].str.lower()
orders['price'] = orders['price'].str.replace('$','',regex = False)
orders['price'] = pd.to_numeric(orders['price'], errors='coerce')
orders['loyalty'] = orders['loyalty'].str.lower()

o = orders.drop_duplicates().copy()

print(o['item'].value_counts().idxmax(),'was ordered the most time')
print(o['size'].value_counts().idxmax(),'was ordered the most time')
print(o.groupby('item')['price'].mean())

print((o['loyalty'] == 'yes').mean(),'used loyalty card')
```

## Level 2 — Employee payroll

An HR export with one duplicate and several cleaning issues. Clean it, then answer:

1. Which department has the highest average salary?
2. What fraction of employees are full-time?
3. Is there a correlation between years of experience and salary? Use `np.corrcoef`.
4. What is the 90th percentile salary across all employees? Use `np.percentile`.
5. Among full-time employees only, which department has the most headcount?

```python
employees = pd.DataFrame({
    'emp_id':    ['E01','E02','E03','E04','E05','E06','E07','E08','E09','E10','E11','E05'],
    'name':      ['Alice Tan','Bob Lee','Carol Wu','David Kim','Eve Park',
                  'Frank Ng','Grace Liu','Henry Chen','Iris Ho','Jack Ma','Karen Yip','Eve Park'],
    'dept':      ['ENGINEERING','Marketing','engineering','Sales','MARKETING',
                  'engineering','Sales','ENGINEERING','marketing','Sales','Engineering','MARKETING'],
    'salary':    ['$95,000','$62,000','$88,000','$54,000','$71,000',
                  '$91,000','$58,000','$102,000','$65,000','$49,000','$84,000','$71,000'],
    'years_exp': [5, 3, 7, 2, 4, 8, 3, 10, 4, 1, 6, 4],
    'full_time': ['TRUE','TRUE','False','TRUE','true','FALSE','TRUE','True','FALSE','TRUE','true','true'],
})

# Your code here

employees['dept'] = employees['dept'].str.lower()
employees['salary'] = employees['salary'].str.replace('$','', regex = False)
employees['salary'] = employees['salary'].str.replace(',','', regex = False)
employees['salary'] = pd.to_numeric(employees['salary'], errors='coerce')
employees['full_time'] = employees['full_time'].str.lower().map({'true': True, 'false': False})
e = employees.drop_duplicates().copy()

print(employees.groupby('dept')['salary'].mean().idxmax(), 'has the highest average salary')
print(employees['full_time'].mean(),'are full time')
cc = np.corrcoef(e['years_exp'],e['salary'])
print('correlation is', cc[0,1])
print('highly correlated')
print(np.percentile(e['salary'],90))
f = e.loc[e['full_time']]
print(f['dept'].value_counts().idxmax(),'has the most headcoutns')
```

## Level 3 — E-commerce orders

Thirteen orders across three categories. One is a duplicate. The `discount` column is stored as strings like `'10%'` — you'll need to strip the `%` and convert before using it.

No steps. Answer these five questions:

1. Clean the data.
2. Add a `final_price` column: `price × (1 - discount / 100)`. Which category has the highest average final price?
3. What fraction of orders have been shipped?
4. What are the 25th and 75th percentile final prices across all orders? Use `np.nanpercentile`.
5. Which customer has spent the most in total? Rank all customers by total spend using `np.argsort`.

```python
ecom = pd.DataFrame({
    'order_id':  ['O001','O002','O003','O004','O005','O006','O007',
                  'O008','O009','O010','O011','O012','O004'],
    'customer':  ['alice@x.com','bob@x.com','alice@x.com','carol@x.com','dave@x.com',
                  'bob@x.com','carol@x.com','eve@x.com','dave@x.com','alice@x.com',
                  'eve@x.com','bob@x.com','carol@x.com'],
    'category':  ['Electronics','Clothing','electronics','CLOTHING','Home',
                  'ELECTRONICS','home','Clothing','HOME','Electronics',
                  'clothing','HOME','CLOTHING'],
    'price':     ['$299','$45','$189','$79','$149','$399','$89',
                  '$55','$199','$249','$65','$129','$79'],
    'discount':  ['10%','0%','15%','5%','0%','20%','10%',
                  '0%','5%','15%','0%','10%','5%'],
    'shipped':   ['Yes','YES','no','Yes','NO','yes','YES',
                  'no','Yes','YES','no','yes','Yes'],
})

# Your code here

ecom['category'] = ecom['category'].str.lower()
ecom['price'] = pd.to_numeric(ecom['price'].str.replace('$','', regex = False))
ecom['discount'] = pd.to_numeric(ecom['discount'].str.replace('%','',regex = False))
ecom['shipped'] = ecom['shipped'].str.lower().map({'yes': True, 'no': False})
ecom  = ecom.drop_duplicates().copy()
ecom['final_price'] = ecom['price']*(1- ecom['discount']/100)
print(ecom.groupby('category')['final_price'].mean().idxmax(),'has the highest avg final price')
print(ecom['shipped'].mean(),'have been shipped')
print(np.nanpercentile(ecom['final_price'],[25,75]))
g = ecom.groupby('customer')['final_price'].sum()
print(g.idxmax(),'has spent the most')
print(g.index[np.argsort(-g)])
```
